```python
numbers = [2, 4, 5, 3, 2, 3]
freq = {}
seen = set()
for num in numbers:
    if num in freq:
        freq[num] += 1
    else:
        freq[num] = 1
for num in numbers:
    if num in seen:
        print("The first repeating element: ", num)
        break
    else:
        seen.add(num)
```

```python
numbers = [2, 5, 3, 2, 4, 1]
freq = {}
for num in numbers:
    if num in freq:
        freq[num] += 1
    else:
        freq[num] = 1
print("First Non-Repeating element", end=" ")
for num in numbers:
    if freq[num] == 1:
        print(freq[num])
        break
```

```python
#Aggregation
class Employee:
    def __init__(self, name):
      self.name = name
class Department:
    def __init__(self, dept_name, Employee):
        self.dept_name = dept_name
        self.Employee = Employee
    def display(self):
        print("Department:", self.dept_name)
        print("Employee:", self.Employee.name)
#Create employee independently
emp = Employee("Aleena")
#Pass employee object to department
dept = Department("IT",emp)
dept.display()
    
```

```python
def factorial(n):
    if n == 0:
        return 1
    else:
        return n * factorial(n - 1)
```

```python
print(factorial(5))
```

```python
#Abstract Method
from abc import ABC, abstractmethod
class Animal(ABC):
    @abstractmethod
    def move(self):
        pass
class Human(Animal):
    def move(self):
        print("I can walk and run")
class Snake(Animal):
    def move (self):
        print("I can crawl")
class Lion(Animal):
    def move(self):
        print("I can roar")
simba = Lion()
simba.move()
```

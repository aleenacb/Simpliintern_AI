```python
numbers = [2, 4, 5, 3, 2, 3]
freq = {}
seen = set()
for num in numbers:
    if num in freq:
        freq[num] += 1
    else:
        freq[num] = 1
for num in numbers:
    if num in seen:
        print("The first repeating element: ", num)
        break
    else:
        seen.add(num)
```

```python
numbers = [2, 5, 3, 2, 4, 1]
freq = {}
for num in numbers:
    if num in freq:
        freq[num] += 1
    else:
        freq[num] = 1
print("First Non-Repeating element", end=" ")
for num in numbers:
    if freq[num] == 1:
        print(freq[num])
        break
```
